```python
%run classifier.ipynb
%run api_handler.ipynb
%run data_exporter.ipynb

def main():
    answer_job = input("Enter a job keyword: ")
    answer_region = input("Which municipality? ")

    answer_jobs = answer_job.strip().lower().split()
    answer_region = answer_region.strip().lower()  # CHANGED

    if len(answer_jobs) == 0:
        print("Please enter at least one keyword.")
        return

    if len(answer_jobs) > 2:
        print("Please enter only one or two keywords.")
        return

    if len(answer_jobs) == 1:
        word = answer_jobs[0]

        if is_architect_term(word):
            question_field = input(
                "Menar du arkitektur inom byggnation eller IT? "
            )
            selected_field = classify_field_answer(question_field)
        else:
            selected_field = classify_search_words(answer_jobs)
    else:
        selected_field = classify_search_words(answer_jobs)

    jobs = fetch_jobs(answer_jobs, answer_region, limit=5)
    print("Jobs fetched from API:", len(jobs))  # CHANGED

    cleaned_jobs = [clean_job(job) for job in jobs]

    filtered_jobs = []

    for job in cleaned_jobs:
        text = (
            job.get("headline", "") + " " +
            job.get("description", "") + " " +
            job.get("employer", "")
        ).lower()

        if ad_contains_all_words(text, answer_jobs):
            filtered_jobs.append(job)

    print("Jobs after keyword filtering:", len(filtered_jobs))  # CHANGED

    result = build_final_output(
        filtered_jobs,
        answer_jobs,
        answer_region,
        selected_field
    )

    print_result(result)


main()
```
