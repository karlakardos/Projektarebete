```python
# MAIN NOTEBOOK to run the program and collect input from the user, then call the functions

%run classifier.ipynb # import the classifier notebook 
%run api_handler.ipynb # import the api_handler notebook 
%run data_exporter.ipynb # import the data_exporter notebook 

def main():
    answer_job = input("Välj en eller två jobbnyckelord: ") # first question to the user to search job

    answer_jobs = answer_job.strip().lower().split() # split the input into a list of keywords, lowercase and remove whitespace

    if len(answer_jobs) == 0:
        print("Vänligen ange minst ett nyckelord.") # empty input handling
        return

    if len(answer_jobs) > 2:
        print("Vänligen ange endast ett eller två nyckelord.") # input handling for more than 2 keywords
        return

    if len(answer_jobs) == 1:
        word = answer_jobs[0] # one keyword input handling

        if is_architect_term(word): # after choosing the word architect/arkitekt, ask for field clarification
            question_field = input(
                "Menar du arkitektur inom bygg eller IT? "
            )
            selected_field = classify_field_answer(question_field) # classifying the IT/bygg answer into a field
        else:
            selected_field = classify_search_words(answer_jobs) # classifying the input keyword into a field for 1 keyword input
    else:
        selected_field = classify_search_words(answer_jobs) # classifying the input keywords into a field for 2 keyword input


# main function to fetch jobs from the API, clean and filter them based on the user's input, and print the results

jobs = fetch_jobs(answer_jobs, limit=5) # fetch 5 jobs from the API based on the user's input keywords

filtered_jobs = [] # empty list to store the filtered jobs

# loop through the fetched jobs and filter them based on the user's input keywords and selected field
for job in jobs:
    text = (
        job.get("headline", "") + " " +
        job.get("description", "") + " " +
        job.get("employer", "") + " " +
        job.get("occupation", "")
    ).lower()

    job_field = classify_job_text(text) # classify the job text into a field (IT, Architecture, or unknown)

    if (
        ad_contains_all_words(text, answer_jobs)
        and job_field == str(selected_field)
    ):
        filtered_jobs.append(job) # append the results into the list
        
# print number of results
       
    print("Jobs after keyword filtering:", len(filtered_jobs))  # print results, number of filtered jobs

    result = build_final_output(
        filtered_jobs,
        answer_jobs,
        selected_field
    )

    print_result(result) # print the final results


main()
```
