```python
from keywords import ARCHITECTURE_KEYWORDS, IT_KEYWORDS


class Job:
    def __init__(self, keywords):
        self.keywords = keywords

    def get_category(self):
        return "okänd"

    def __str__(self):
        return self.get_category()


class ITJob(Job):
    def get_category(self):
        return "IT"


class ArchitectureJob(Job):
    def get_category(self):
        return "byggnation/arkitektur"


def is_architect_term(word):
    cleaned_word = word.strip().lower()
    return cleaned_word in {"arkitekt", "arkitektur"}


def classify_field_answer(answer):
    answer_words = answer.strip().lower().split()

    if any(word in IT_KEYWORDS for word in answer_words):
        return ITJob(answer_words)

    if any(word in ARCHITECTURE_KEYWORDS for word in answer_words):
        return ArchitectureJob(answer_words)

    return Job(answer_words)


def ad_contains_all_words(ad_text, search_words):
    text = ad_text.lower()
    for word in search_words:
        if word.strip().lower() not in text:
            return False
    return True

def classify_search_words(words):
    cleaned_words = [word.strip().lower() for word in words]

    # One-word logic
    if len(cleaned_words) == 1:
        has_it_keyword = any(word in IT_KEYWORDS for word in cleaned_words)
        has_architecture_keyword = any(
            word in ARCHITECTURE_KEYWORDS for word in cleaned_words
        )

        if has_it_keyword and has_architecture_keyword:
            return Job(cleaned_words)
        if has_it_keyword:
            return ITJob(cleaned_words)
        if has_architecture_keyword:
            return ArchitectureJob(cleaned_words)
        return Job(cleaned_words)

    # Two-word case: keep the previous requirement structure
    if len(cleaned_words) == 2:
        return Job(cleaned_words)

    return Job(cleaned_words)
```
