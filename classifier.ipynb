```python
# CLASSESIFIER NOTEBOOK to classify the job keywords and job text

# first sorting of the keywords into IT, Architecture and Role keywords

from keywords import ARCHITECTURE_KEYWORDS, IT_KEYWORDS, ROLE_KEYWORDS # import of keywords

def is_architect_term(word): # clean the input word
    cleaned_word = word.strip().lower()
    return cleaned_word in {"arkitekt", "arkitektur", "architect", "architecture"}

# parent class
class Job: # job classification
    def __init__(self, keywords):
        self.keywords = keywords

    def get_category(self): # when input is not classified, unknown
        return "okänd"

    def __str__(self): # creating string from keywords and category
        return self.get_category()

# child class 1
class ITJob(Job): # for IT job classification
    def get_category(self):
        return "IT"

# child class 2
class ArchitectureJob(Job): # for Architecture job classification
    def get_category(self):
        return "bygg/arkitektur"


# function takes the answer from the user and classifies it into IT, Architecture or unknown
def classify_field_answer(answer):
    answer_words = answer.strip().lower().split()

    if any(word in IT_KEYWORDS for word in answer_words): # returning IT jobs if words are in IT list
        return ITJob(answer_words)

    if any(word in ARCHITECTURE_KEYWORDS for word in answer_words): # returning Architecture jobs if words are in Architecture list
        return ArchitectureJob(answer_words)

    return Job(answer_words) # returning unknown jobs if words are not in any list
```

```python
# sorting of answers with two keywords into IT, Architecture or unknown

def ad_contains_all_words(ad_text, search_words): 
    text = ad_text.lower()
    for word in search_words:
        if word.strip().lower() not in text:
            return False
    return True


def classify_job_text(text):
    text = text.lower()

    has_it_keyword = any(
        word in text for word in IT_KEYWORDS
    )

    has_architecture_keyword = any(
        word in text for word in ARCHITECTURE_KEYWORDS
    )

    if has_it_keyword and not has_architecture_keyword:
        return "IT"

    if has_architecture_keyword and not has_it_keyword:
        return "bygg/arkitektur"

    return "okänd"

    
def classify_search_words(words):
    cleaned_words = [word.strip().lower() for word in words]

    if len(cleaned_words) == 1:
        if cleaned_words[0] in ROLE_KEYWORDS:
            return Job(cleaned_words)  # ask the user which field

        has_it_keyword = any(word in IT_KEYWORDS for word in cleaned_words)
        has_architecture_keyword = any(
            word in ARCHITECTURE_KEYWORDS for word in cleaned_words
        )

        if has_it_keyword and has_architecture_keyword:
            return Job(cleaned_words)
        if has_it_keyword:
            return ITJob(cleaned_words)
        if has_architecture_keyword:
            return ArchitectureJob(cleaned_words)
        return Job(cleaned_words)

    if len(cleaned_words) == 2:
        role_matches = sum(word in ROLE_KEYWORDS for word in cleaned_words)
        it_matches = sum(word in IT_KEYWORDS for word in cleaned_words)
        architecture_matches = sum(
            word in ARCHITECTURE_KEYWORDS for word in cleaned_words
        )

        if role_matches == 1 and it_matches:
            return ITJob(cleaned_words)

        if role_matches == 1 and architecture_matches:
            return ArchitectureJob(cleaned_words)

        if it_matches == 2:
            return ITJob(cleaned_words)

        if architecture_matches == 2:
            return ArchitectureJob(cleaned_words)

        if it_matches and architecture_matches:
            return Job(cleaned_words)

        return Job(cleaned_words)

    return Job(cleaned_words)
```
