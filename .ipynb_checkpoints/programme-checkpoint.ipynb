```python
from client import Client
from calculation import Calculation_BMI
from client_data import client_data


def input_client():
    name = input("Enter your name: ")
    age = int(input("Enter your age: "))
    weight = float(input("Enter your weight (kg): "))
    height = float(input("Enter your height (cm): "))
    label = int(input("Enter your label (1 for healthy, 0 for mild healthy issue): "))
    new_client = [name, age, [weight, height], label]
    client_data.append(new_client)
    return new_client


input_client()

for data in client_data:
    Client(*data)

for client in Client.all_clients:
    print(client.name)

calculation = Calculation_BMI()
for client in Client.all_clients:
    BMI = calculation.calculate_BMI(client.measurements)
    print(f"{client.name}, {client.get_gender()}, has {BMI:.2f} BMI.")
```
